```python
import numpy as np
import tensorflow as tf

SEED = 42
np.random.seed(SEED)
tf.random.set_seed(SEED)
```

```python
from tensorflow import keras
from tensorflow.keras.preprocessing.image import load_img
```

```python
path = './data/train/curly'
name = '2a6d1e3e1c692ca898e3d3a75fd5043b4.jpg'
fullname = f'{path}/{name}'
load_img(fullname)
```

```python
inputs = keras.Input(shape=(200, 200, 3))
conv = keras.layers.Conv2D(
    filters=32,
    kernel_size=(3,3),
    activation='relu',
)(inputs)
featMap = keras.layers.MaxPooling2D(pool_size=(2,2))(conv)
vectors = keras.layers.Flatten()(featMap)
inner = keras.layers.Dense(64, activation='relu')(vectors)
outputs = keras.layers.Dense(1, activation='sigmoid')(inner)
model = keras.Model(inputs, outputs)

optimizer = keras.optimizers.SGD(learning_rate=0.002, momentum=0.8)

model.compile(
    optimizer=optimizer,
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model.summary()
```
